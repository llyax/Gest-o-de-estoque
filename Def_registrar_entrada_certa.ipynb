```python
def registrar_entrada(produto, quantidade, estoque):
  produto = produto.strip()
  if not produto:
    print("Erro: o nome do produto não pode estar vazio!")
    return False
  if quantidade <= 0:
    print("Erro: a quantidade deve ser maior que zero!")
    return False
    if produto in estoque:
      estoque[produto][0] += quantidade
      print(f"Entrada registrada com sucesso!")
      print(f"Novo estoque de '{produto}': {int(estoque[produto][0])} unidades.")
    return True
  else:
    print("Produto não encontrado no estoque.")
    return False
```
