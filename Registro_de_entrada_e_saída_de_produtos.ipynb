```python
def registrar_entrada(produto: str, quantidade: int, estoque: dict) -> bool:

    produto = produto.strip().upper()

    if not produto:
        print("Erro: o nome do produto não pode estar vazio.")
        return False

    if not isinstance(quantidade, int) or isinstance(quantidade, bool):
        print("Erro: a quantidade deve ser um número inteiro.")
        return False

    if quantidade <= 0:
        print("Erro: a quantidade deve ser maior que zero.")
        return False

    estoque[produto] = estoque.get(produto, 0) + quantidade
    print(f"Sucesso: {quantidade} unidade(s) de '{produto}' registrada(s). "
          f"Total em estoque: {estoque[produto]}")
    return True
```
